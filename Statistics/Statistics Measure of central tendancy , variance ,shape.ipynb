```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
data = np.array([30 , 32 , 35 , 36 , 38 , 40 ,45 , 100])
```

```python
# # measures of central tendacy 
# 1 - mean 
# 2 - median 
# 3 - mode
```

```python
# Mean ( Average )
mean_value = np.mean(data)
mean_value
```

```python
# Meadian ( Middle value after sorting )

median_value = np.median(data)
median_value
```

```python
# Mode ( mostly frequent )

from scipy import stats

mode_value = stats.mode(data, keepdims = True )
mode_value
```

```python
# # lets start measure of varibility 
# Main types are 
# 1 - range 
# 2 - variance 
# 3 - standerd deviation 
```

```python
# range 
range_value = np.max(data) - np.min(data)
range_value
```

```python
# variance 
variance_value = np.var(data)
variance_value
```

```python
# standerd deviation 
std_dev_value = np.std(data)
std_dev_value
```

```python
# # Measures of shape ( it tells how the data looks )
# main types 
# 1 - skewness 
# 2 kurtosis 
```

```python
# 1 - skewness 

plt.hist(data,bins=6)
plt.title("histogram of data skewness")
plt.xlabel("value")
plt.ylabel("frequency")
plt.show()
```

```python
# 2 - Kurtosis 
kurtosis_value = stats.kurtosis(data)
kurtosis_value
```
