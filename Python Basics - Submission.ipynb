```python
# 5.2. Set Operations Given two sets, write a Python program to find the union, intersection,
# and difference of these sets."

# Create two sets
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Union (all unique elements from both sets)
union_set = set1.union(set2)

# Intersection (common elements)
intersection_set = set1.intersection(set2)

# Difference (elements in set1 but not in set2)
difference_set = set1.difference(set2)

# Print results
print("Set1:", set1)
print("Set2:", set2)
print("Union:", union_set)
print("Intersection:", intersection_set)
print("Difference (set1 - set2):", difference_set)
```

# 5. Explain the concept of Sets and their properties in Python. Provide a one-line definition
# of each set method with examples.

A set in Python is a collection of unique elements stored in unordered form and defined using {}.

👉 Example:

my_set = {1, 2, 3, 4}
📌 Properties of Set
Unordered
Elements do not have fixed positions (no indexing)
Unique Elements
Duplicate values are automatically removed
Mutable
You can add or remove elements
No Indexing / Slicing
Cannot access elements using index
Heterogeneous
Can store different data types

```python
# Initial sets
s = {1, 2, 3}
s1 = {1, 2}
s2 = {2, 3, 4}

# add() → Adds a single element
s.add(4)
print("add:", s)

# update() → Adds multiple elements
s.update([5, 6])
print("update:", s)

# remove() → Removes element (error if not found)
s.remove(1)
print("remove:", s)

# discard() → Removes element (no error if not found)
s.discard(10)
print("discard:", s)

# pop() → Removes random element
s.pop()
print("pop:", s)

# copy() → Creates a copy
new_s = s.copy()
print("copy:", new_s)

# clear() → Removes all elements
temp = s.copy()
temp.clear()
print("clear:", temp)

# union() → All elements from both sets
print("union:", s1.union(s2))

# intersection() → Common elements
print("intersection:", s1.intersection(s2))

# difference() → Elements in s1 but not in s2
print("difference:", s1.difference(s2))

# symmetric_difference() → Not common elements
print("symmetric_difference:", s1.symmetric_difference(s2))

# intersection_update() → Updates s1 with common elements
temp1 = s1.copy()
temp1.intersection_update(s2)
print("intersection_update:", temp1)

# difference_update() → Updates s1 removing common elements
temp2 = s1.copy()
temp2.difference_update(s2)
print("difference_update:", temp2)

# symmetric_difference_update() → Updates with non-common elements
temp3 = s1.copy()
temp3.symmetric_difference_update(s2)
print("symmetric_difference_update:", temp3)

# issubset() → Check if s1 is subset of s2
print("issubset:", s1.issubset(s2))

# issuperset() → Check if s2 is superset of s1
print("issuperset:", s2.issuperset(s1))

# isdisjoint() → Check if no common elements
print("isdisjoint:", s1.isdisjoint(s2))
```

1. Define Dictionary, its properties and examples for each method

A dictionary in Python is a collection of key-value pairs where each key is unique and used to access its corresponding value. It is defined using curly braces {}.

propeerties:

Ordered (Python 3.7+)
Elements are stored in insertion order
Mutable
Values can be added, updated, or deleted
Keys are Unique
Duplicate keys are not allowed
Values can be Duplicate
Same value can appear multiple times
Keys must be Immutable
Example: string, number, tuple

```python
# Methods with examples

# Initial dictionary
d = {"name": "Arjun", "age": 21}

# get() → returns value of key
print("get:", d.get("name"))

# keys() → returns all keys
print("keys:", d.keys())

# values() → returns all values
print("values:", d.values())

# items() → returns key-value pairs
print("items:", d.items())

# update() → updates dictionary
d.update({"age": 22, "city": "Pune"})
print("update:", d)

# pop() → removes key and returns value
d.pop("age")
print("pop:", d)

# popitem() → removes last inserted item
d.popitem()
print("popitem:", d)

# add new key using assignment
d["country"] = "India"
print("add:", d)

# copy() → creates copy
new_d = d.copy()
print("copy:", new_d)

# clear() → removes all elements
temp = d.copy()
temp.clear()
print("clear:", temp)

# fromkeys() → creates new dictionary with given keys
keys = ["a", "b", "c"]
new_dict = dict.fromkeys(keys, 0)
print("fromkeys:", new_dict)

# setdefault() → returns value, if not present adds key
d.setdefault("gender", "male")
print("setdefault:", d)
```

```python
# 5. Create a Python program that stores the grades of five students in a dictionary. Allow
# users to input the names of the students and their respective grades. Calculate and
# display the average grade of the students.

# Create empty dictionary
students = {}

# Take input for 5 students
for i in range(5):
    name = input("Enter student name: ")
    grade = float(input("Enter grade: "))
    students[name] = grade

# Print dictionary
print("Student Grades:", students)

# Calculate average
total = sum(students.values())
average = total / len(students)

# Display average
print("Average Grade:", average)
```

```python
# 4. Construct your own function my_range() which acts as range() to print numbers in
# forward direction (step is positive) and backward direction(step is negative) with given
# start,stop and step as parameters.
# eg: my_range(1,10,1), then o/p= 1,2,3,4,5,6,7,8,9
# my_range(1,10,2), then o/p= 1,3,5,7,9
# my_range(10,1,-1), then o/p = 10,9,8,7,6,5,4,3,2
# my_range(10,1,-2), then o/p = 10,8,6,4,2

def my_range(start, stop, step):
    if step == 0:
        print("Step cannot be zero")
        return

    # Forward direction
    if step > 0:
        i = start
        while i < stop:
            print(i, end=",")
            i += step

    # Backward direction
    else:
        i = start
        while i > stop:
            print(i, end=",")
            i += step


# Test cases
my_range(1, 10, 1)
print()

my_range(1, 10, 2)
print()

my_range(10, 1, -1)
print()

my_range(10, 1, -2)
```

```python
# 5. Multiply numbers with given start and stop (using loops)
#  Example: start=1, stop= 5, then o/p: 120
#  start=3, stop=5, th


def multiply_range(start, stop):
    result = 1

    for i in range(start, stop + 1):
        result *= i

    return result


# Example usage
print(multiply_range(1, 5))  # 120
print(multiply_range(3, 5))  # 60
```

```python
# 6. Construct a function for fibonacci series(fib()) using
#  a.loops
#  b.recursion
#  Expected o/p :
# fib(5), then o/p= 0 1 1 2 3
# fib(7), then o/p= 0 1 1 2 3 5 8

def fib(n):
    a, b = 0, 1

    for _ in range(n):
        print(a, end=" ")
        a, b = b, a + b


# Example
fib(5)
print()
fib(7)
```

```python
def fib_rec(n):
    if n <= 1:
        return n
    return fib_rec(n-1) + fib_rec(n-2)


def fib(n):
    for i in range(n):
        print(fib_rec(i), end=" ")


# Example
fib(5)
print()
fib(7)
```

```python
# 7. Create a Factorial with loops.

def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result *= i

    return result


# Example
print(factorial(5))  # 120
print(factorial(3))  # 6
```

```python
# Task 11 - Problem Solving:
# 1. Write a function called "count_vowels" that takes a string as input and returns the
# number of vowels (a, e, i, o, u) in that string.
# 2. Write a Python function to sum all the numbers in a list.(Without using built-in function
# sum).
# 3. Write a Python function to check whether a number falls within a given range. The
# function should take three arguments: start, stop, and number. The function should
# return True if the number is within the range [start, stop] (inclusive), and False
# otherwise.
# 4. Write a Python function that accepts a string and counts the number of upper and lower
# case letters
# 5. Write a function is_prime(n) that returns True if the number n is prime, otherwise False.
# 6. Write a function is_anagram(s1, s2) that takes two strings s1 and s2 and returns True if
# they are anagrams of each other, otherwise False.
# is_anagram("listen", "silent") # Output: True
# is_anagram("hello", "world") # Output: False
# 7. Write a function sum_of_digits(n) that takes an integer n and returns the sum of its
# digits.
# sum_of_digits(1234) # Output: 10


# Task 11 - Problem Solving

# 1. Count vowels in a string
def count_vowels(s):
    count = 0
    for ch in s.lower():
        if ch in "aeiou":
            count += 1
    return count


# 2. Sum of all numbers in a list (without sum())
def list_sum(lst):
    total = 0
    for num in lst:
        total += num
    return total


# 3. Check if number is within range
def in_range(start, stop, number):
    return start <= number <= stop


# 4. Count uppercase and lowercase letters
def count_case(s):
    upper = 0
    lower = 0

    for ch in s:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1

    return upper, lower


# 5. Check prime number
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False

    return True


# 6. Check anagram
def is_anagram(s1, s2):
    return sorted(s1) == sorted(s2)


# 7. Sum of digits
def sum_of_digits(n):
    total = 0
    for digit in str(n):
        total += int(digit)
    return total


# ------------------ TEST CASES ------------------

print("1:", count_vowels("Hello World"))           # 3
print("2:", list_sum([1, 2, 3, 4, 5]))             # 15
print("3:", in_range(1, 10, 5))                    # True
print("4:", count_case("Hello World"))             # (2, 8)
print("5:", is_prime(7))                           # True
print("6:", is_anagram("listen", "silent"))        # True
print("7:", sum_of_digits(1234))                   # 10
```

```python
# Task 14 - OOPS

# 1. Vehicle class with static + instance attributes
class Vehicle:
    max_speed = 240   # static variable

    def __init__(self, mileage):
        self.mileage = mileage   # instance variable

v = Vehicle(18)
print("max_speed =", Vehicle.max_speed)
print("mileage =", v.mileage)


# --------------------------------------------------

# 2. Car class with attributes
class Car:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    def display_info(self):
        print("Car:", self.make, self.model)

c1 = Car("Toyota", "Innova")
print(c1.make, c1.model)
c1.display_info()


# --------------------------------------------------

# 3. Constructor already used above (✔)


# --------------------------------------------------

# 4. Count total cars
class Car:
    total_cars = 0

    def __init__(self, make, model):
        self.make = make
        self.model = model
        Car.total_cars += 1

    def get_total_cars(self):
        return Car.total_cars

c1 = Car("Toyota", "Innova")
c2 = Car("Honda", "City")
print("Total Cars:", c1.get_total_cars())


# --------------------------------------------------

# 5. Arithmetic class
class Calculator:
    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        return a / b

calc = Calculator()
print(calc.add(10, 5))
print(calc.subtract(10, 5))


# --------------------------------------------------

# 6. Inheritance (Vehicle -> Car)
class Vehicle:
    def __init__(self, color, model):
        self.color = color
        self.model = model

class Car(Vehicle):
    def __init__(self, color, model, num_doors):
        super().__init__(color, model)
        self.num_doors = num_doors

c = Car("Red", "MG Hector", 4)
print(f"Car: {c.model}, {c.color}; Doors: {c.num_doors}")


# --------------------------------------------------

# 7. Multiple Inheritance
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Address:
    def __init__(self, city, country):
        self.city = city
        self.country = country

class Employee(Person, Address):
    def __init__(self, name, age, city, country, job):
        Person.__init__(self, name, age)
        Address.__init__(self, city, country)
        self.job = job

e = Employee("Sirisha", 25, "Delhi", "India", "Data Scientist")
print(f"Employee: {e.name}, Age: {e.age}, City: {e.city}, Country: {e.country}, Job: {e.job}")


# --------------------------------------------------

# 8. Private attributes with getter/setter
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_age(self):
        return self.__age

    def set_age(self, age):
        self.__age = age

p = Person("Arjun", 21)
print(p.get_name(), p.get_age())


# Using properties
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    @property
    def name(self):
        return self.__name

    @name.setter
    def name(self, value):
        self.__name = value

p = Person("Arjun", 21)
print(p.name)


# --------------------------------------------------

# 9. Protected + module-like example
class BankAccount:
    def __init__(self, balance):
        self._balance = balance   # protected

    def get_balance(self):
        return self._balance

acc = BankAccount(5000)
print("Balance:", acc.get_balance())


# Simulating module
def __add(a, b):
    return a + b

def add_numbers(a, b):
    return __add(a, b)

print(add_numbers(3, 4))


# --------------------------------------------------

# 10. Abstract class
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def calculate_area(self):
        pass

class Rectangle(Shape):
    def __init__(self, l, w):
        self.l = l
        self.w = w

    def calculate_area(self):
        return self.l * self.w

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def calculate_area(self):
        return 3.14 * self.r * self.r

r = Rectangle(5, 3)
c = Circle(2)

print("Rectangle Area:", r.calculate_area())
print("Circle Area:", c.calculate_area())
```
