## 1. Dictionaries in Python

A **dictionary** stores data in **key : value pairs**.

Example:

student = {"name":"Rahul", "age":22, "course":"Data Science"}

Important Dictionary Methods:
- keys()
- values()
- items()


## 2. Functions in Python
