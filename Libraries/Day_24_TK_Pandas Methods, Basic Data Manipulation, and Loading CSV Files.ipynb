# 9️⃣ Filtering Rows with Conditions
#### 👉 Single condition


```python
# I want all female whose age is more than 30 
df[df['Age'] > 30]
df[df['Sex']=='female']
```

#### 👉 Multiple conditions

```python
# find all peoples whose age is greater than 35 and fare is greater than 50
df[(df['Age'] > 35) & (df['Fare'] > 50)]
```

## ⚠️ Important:

Use &, |

Always wrap conditions in brackets

# 1️⃣2️⃣ Missing Values (VERY IMPORTANT)

```python
#### 👉 Checking missing values
- df.isnull()
- df.isnull().sum()
- df.notnull()

#### 👉 Dropping missing values
#- df.dropna()
#- df.dropna(axis=1)
```

```python
df.isnull().sum()
```

#### 👉 Filling missing values

```python
# Way 1 --> replacing by 0 
df=df.fillna(0)
df.head()
```

```python
df.isnull().sum()
```

```python
import warnings 
warnings.filterwarnings("ignore")

# Way 2 (restart kernel and replace one by one)

df['Age'].fillna(df['Age'].mean(),inplace = True)

df["Cabin"].fillna(df["Cabin"].mode()[0] , inplace = True)

df['Fare'].fillna(df['Fare'].mean() , inplace = True)
```

```python
df.isnull().sum()
```

#### 🧠 Interview Note:

- Mean → numerical
- Mode → categorical
