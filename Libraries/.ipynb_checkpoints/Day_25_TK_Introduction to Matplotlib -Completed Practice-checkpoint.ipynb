## 1. What is Matplotlib?
Matplotlib is a **Python library for data visualization** used to create plots, charts, and graphs.

### Why Matplotlib is important in Data Analytics:
- Understand trends & patterns
- Compare categories
- Analyze distributions
- Detect outliers
- Core library used during **EDA (Exploratory Data Analysis)**

## Sample Dataset (Used for All Examples)

``````python
import pandas as pd 
data= {
    'Day': [1,2,3,4,5,6,7],
    'Sales': [250,300,280,350,400,420,390],
    'Profit': [50,70,60,90,110,130,100],
    'Category': ['A','B','A','C','B','C','A']
}

df = pd.DataFrame(data)
print(df)
``````

## Basic Plot Syntax


```python
plt.plot(x, y)
plt.show()
```

``````python
# Display day & sales in plot
plt.plot(df['Day'], df['Sales'])
plt.show()
``````

## Line Plot – Trend Over Time
```Used to visualize continuous data over time or sequence.```

``````python
#Display day wise sales

plt.figure(figsize=(6,4))
plt.plot(df['Day'], df['Sales'], marker='o', color='blue')
plt.title('Sales Trend')
plt.xlabel('Day')
plt.ylabel('Sales')
plt.show()
``````

## Bar Plot – Category Comparison

Used for **comparing categories**


``````python
import matplotlib.pyplot as plt
plt.bar(df['Day'], df['Sales'], color='orange')
plt.title('Sales Comparison')
plt.xlabel('Day')
plt.ylabel('Sales')
plt.show()
``````

## Histogram – Data Distribution

Used to visualize **distribution of numerical data**.

``````python
import matplotlib.pyplot as plt
plt.hist(df['Sales'], bins=5)
plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
``````

## Box Plot – Detecting Outliers

Used to identify **spread and outliers**.


`````Outliers are data points that are significantly different from the rest of the values in a dataset. They may occur due to measurement errors, data entry mistakes, or genuine extreme observations. Outliers can strongly affect statistical measures like mean, standard deviation, and model performance, so identifying and handling them is an important step in data analysis and machine learning.`````

``````python
# In a list of ages [22, 24, 25, 23, 120], the value 120 is an outlier.
# If most students score between 40–90 in an exam and one student scores 5, then 5 is an outlier.
# Salaries [30k, 32k, 35k, 33k, 300k] → 300k is an outlier.
# Daily website visits [100, 120, 110, 115, 900] → 900 is an outlier.
``````

``````python
plt.boxplot(df['Profit'])
plt.title('Profit Distribution')
plt.ylabel('Profit')
plt.show()
``````

## Scatter Plot – Relationship Between Variables
Used to understand **relationship between two variables**.

``````python
plt.scatter(df['Sales'], df['Profit'])
plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
``````

## Multiple Lines in One Plot (Interview Important)


``````python
plt.plot(df['Day'], df['Sales'], marker='o', label='Sales')
plt.plot(df['Day'], df['Profit'], marker='s', label='Profit')
plt.title('Sales & Profit Trend')
plt.xlabel('Day')
plt.ylabel('Amount')plt.plot(df)
plt.legend()
plt.show()
``````

``````python
# # B. Coding Practice

# 1.Create a line plot for temperature vs days
import pandas as pd 

dataa = {
   'Temprature' : [35,40,45,50,55,60],
    'Days'      : [1,2,3,4,5,6]
  
}

df =pd.DataFrame(dataa)
print(df)
``````

``````python
import matplotlib.pyplot as plt

plt.figure(figsize=(4,4))
plt.plot(df['Days'], df['Temprature'], marker='o', color='blue')
plt.title('Trend')
plt.xlabel('Day')
plt.ylabel('Temprature')
plt.show()
``````

``````python
# 2. Create a bar chart showing students count per batch

import pandas as pd 

school = {
   'Count': [35,40,45,50,55,60],
    'Batch' : [1,2,3,4,5,6]
  
}

df =pd.DataFrame(school)
print(df)
``````

``````python
import matplotlib.pyplot as plt

plt.bar(df['Batch'], df['Count'], color='orange')
plt.title('Sales Comparison')
plt.xlabel('Batch')
plt.ylabel('Student Count')
plt.show()

``````

``````python
# 3 .Plot a histogram for randomly generated data

import numpy as np
import matplotlib.pyplot as plt

# random data generate 
data = np.random.randn(1000)   # 1000 random values

# histogram plot 
plt.hist(data, bins=30, color='blue')

plt.title('Histogram of Random Data')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.show()
``````

``````python
import matplotlib.pyplot as plt

# sample data
height = [150, 155, 160, 165, 170, 175, 180]
weight = [50, 55, 60, 65, 70, 75, 80]

# scatter plot
plt.scatter(height, weight, color='green')

plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')

plt.show()
``````

# 📊 Interview-Oriented Questions (EDA & Matplotlib)

---

## 1. Why is visualization important in EDA?

Visualization helps to quickly understand data patterns, trends, relationships, and anomalies.
It makes complex data easier to interpret and supports better decision-making.

---

## 2. Can Matplotlib be used with Pandas DataFrames?

Yes, Matplotlib works seamlessly with Pandas DataFrames.
You can directly use DataFrame columns for plotting.

**Example:**

```python
df.plot()   # Uses matplotlib internally
```

```python
import matplotlib.pyplot as plt
plt.plot(df['col1'], df['col2'])
```

---

## 3. How do you save a plot?

You can save a plot using `plt.savefig()`.

**Example:**

```python
plt.savefig('plot.png')
```

You can save in different formats like PNG, JPG, or PDF.

---

## 4. What happens if `plt.show()` is not used?

In scripts, the plot may not be displayed if `plt.show()` is not used.
In Jupyter Notebook, plots may still appear automatically, but it is best practice to include it.

---

## 5. Which plot is best to detect outliers and why?

A **Box Plot** is best for detecting outliers because it clearly shows:

* Median
* Quartiles (Q1, Q3)
* Extreme values (outliers)

Outliers appear as individual points beyond the whiskers.
